# 03 - Recomendações

**Inferência batch:** o modelo `@champion` calcula o top-10 de **todos** os usuários de
uma vez e grava numa tabela. O app só faz um `SELECT` - rápido e barato. É assim que
muitos portais montam o "recomendado para você". Para o **leitor novo**, gravamos também os
vizinhos de cada matéria (`gold_similares`).

Depois, um **monitoramento** simples: estamos recomendando matérias variadas ou
sempre as mesmas?

```python
# Parâmetro do Job (rodando à mão, fica o valor padrão)
dbutils.widgets.text("schema", "news_recsys_dev")
SCHEMA = f"workspace.{dbutils.widgets.get('schema')}"
print(SCHEMA)
```

```python
import mlflow
import pandas as pd
from mlflow import MlflowClient

mlflow.set_registry_uri("databricks-uc")
MODELO = f"{SCHEMA}.news_recommender"

versao = MlflowClient().get_model_version_by_alias(MODELO, "champion").version
modelo = mlflow.pyfunc.load_model(f"models:/{MODELO}@champion")
print(f"usando {MODELO} versão {versao} (@champion)")
```

```python
usuarios = spark.table(f"{SCHEMA}.silver_users").toPandas()
materias = spark.table(f"{SCHEMA}.bronze_articles").toPandas()

recs = modelo.predict(usuarios[["user_id"]].assign(lidas=""))  # lidas vazio = leitor conhecido
gold = recs.merge(materias[["url", "title", "section", "image_url"]], on="url").assign(
    model_version=str(versao), generated_at=pd.Timestamp.now()
)
spark.createDataFrame(gold).write.mode("overwrite").saveAsTable(f"{SCHEMA}.gold_recommendations")
print(len(gold), "recomendações para", gold["user_id"].nunique(), "usuários")
gold.head(10)[["user_id", "rank", "section", "title"]]
```

### Vizinhos de cada matéria (leitor novo, em batch)

O mesmo `predict`, agora com um **perfil de uma matéria só** para cada matéria do catálogo: o
top-10 de quem leu só aquela matéria. O app junta os vizinhos das matérias que o leitor marcou
(aba *Monte seu perfil (Tabela Gold)*). A aba *Serving Endpoints* manda o perfil inteiro de uma vez.

```python
vizinhos = modelo.predict(pd.DataFrame({"user_id": materias["url"], "lidas": materias["url"]}))
similares = vizinhos.rename(columns={"user_id": "lida"}).assign(model_version=str(versao))
spark.createDataFrame(similares).write.mode("overwrite").saveAsTable(f"{SCHEMA}.gold_similares")
print(len(similares), "vizinhos para", similares["lida"].nunique(), "matérias")
```

## Monitoramento

- **Cobertura:** que fração do catálogo aparece em alguma recomendação?
- **Concentração por leitor:** que % do que cada pessoa lê está na editoria preferida dela? E
  do que ela recebe? Se a recomendação concentra mais do que a leitura, o modelo está criando
  uma **bolha**.
- **Editorias lidas × recomendadas:** a mesma pergunta, olhando o total.

```python
cliques = spark.table(f"{SCHEMA}.silver_clicks").toPandas().merge(materias[["url", "section"]], on="url")
lidas = cliques["section"].value_counts(normalize=True)
recomendadas = gold["section"].value_counts(normalize=True)
cobertura = gold["url"].nunique() / len(materias)


def concentracao(df):  # % médio na editoria preferida de cada leitor
    return df.groupby("user_id")["section"].value_counts(normalize=True).groupby("user_id").max().mean()


concentracao_lida, concentracao_recebida = concentracao(cliques), concentracao(gold)

spark.createDataFrame(pd.DataFrame([{
    "generated_at": pd.Timestamp.now(), "model_version": str(versao),
    "cobertura": cobertura, "usuarios": gold["user_id"].nunique(),
    "concentracao_lida": concentracao_lida, "concentracao_recebida": concentracao_recebida,
}])).write.mode("append").saveAsTable(f"{SCHEMA}.model_monitoring")

print(f"cobertura do catálogo: {cobertura:.0%}")
print(f"na editoria preferida: {concentracao_lida:.0%} do que lê - {concentracao_recebida:.0%} do que recebe")
comparativo = pd.DataFrame({"lidas": lidas, "recomendadas": recomendadas}).fillna(0)
comparativo.sort_values("lidas").tail(10).plot.barh(
    figsize=(8, 4), title="Editorias: lidas × recomendadas (as 10 mais lidas)"
);
```

### O que percebemos?

- A inferência usa o **alias** `@champion`, nunca "a última versão": o quality gate
  protege o app.
- Cada linha da tabela gold diz **qual versão** do modelo a gerou.
- **Bolha de editoria:** a recomendação fica tão presa à editoria preferida quanto a própria
  leitura (~80%). O modelo **não amplia** a bolha, mas também não a quebra.
- **Bolha de catálogo:** só ~7% das matérias aparecem em alguma recomendação. O modelo
  concentra em poucas matérias (populares e recentes). O NDCG sozinho não mostra isso.
